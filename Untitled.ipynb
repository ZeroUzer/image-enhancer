```python
def apply_coeffs(img, coeffs):
    """
    Применяет коэффициенты яркости, контраста и насыщенности к изображению
    coeffs: [k_brightness, k_contrast, k_saturation] в нормализованном виде (0-1)
    """
    # Денормализуем коэффициенты
    k_brightness = 0.5 + coeffs[0] * 1.5
    k_contrast = 0.5 + coeffs[1] * 1.5
    k_saturation = coeffs[2] * 2.0
    
    # Копируем изображение
    result = img.copy().astype(np.float32)
    
    # Применяем яркость и контраст
    result = result * k_contrast + (128 / 255.0) * (1 - k_contrast) + (k_brightness - 1) * 0.5
    result = np.clip(result, 0, 1)
    
    # Применяем насыщенность (через HSV)
    hsv = cv2.cvtColor((result * 255).astype(np.uint8), cv2.COLOR_RGB2HSV).astype(np.float32)
    hsv[:, :, 1] = hsv[:, :, 1] * k_saturation
    hsv[:, :, 1] = np.clip(hsv[:, :, 1], 0, 255)
    result = cv2.cvtColor(hsv.astype(np.uint8), cv2.COLOR_HSV2RGB) / 255.0
    
    return result
```
